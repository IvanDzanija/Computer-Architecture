# Visual comparison of speed - CPU vs GPU (f32-gemm)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Matrix sizes
n_sizes = [128, 256, 512, 1024, 2048]

# Times in ms
cpu_times_ms = [13.2, 71.2, 472.2, 3853.7, 43479.8] 
gpu_times_ms = [0.0023, 0.0234, 0.2826, 0.6189, 2.35] 

plt.figure(figsize=(10, 6))

plt.plot(n_sizes, cpu_times_ms, 'o-', label='CPU Naive (C)', color='red', linewidth=2)
plt.plot(n_sizes, gpu_times_ms, 's-', label='GPU OpenCL (M2 Air)', color='blue', linewidth=2)
plt.yscale('log')
plt.xscale('log', base=2)

plt.title('Usporedba vremena izvođenja: CPU vs GPU (Log skala)', fontsize=14)
plt.xlabel('Veličina matrice (N x N)', fontsize=12)
plt.ylabel('Vrijeme izvođenja (ms)', fontsize=12)
plt.xticks(n_sizes, n_sizes) 
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()

speedup_1024 = cpu_times_ms[3] / gpu_times_ms[3]
plt.annotate(f'Speedup @1024: ~{int(speedup_1024)}x', 
             xy=(1024, gpu_times_ms[3]), xytext=(1024, gpu_times_ms[3]*10),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5))

plt.tight_layout()
plt.show()
```
